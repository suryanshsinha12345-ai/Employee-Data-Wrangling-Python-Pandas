```python
# Import required libraries
import pandas as pd  # Import pandas for DataFrame operations
import numpy as np  # Import numpy for handling missing values

# ==========================================
# Task 1 — Create Data frame and save as CSV
# ==========================================

# Create Project DataFrame data dictionary
project_data = {
    "ID": ["A001","A002","A003","A004","A005","A002","A005","A003","A001","A003","A001","A004","A004","A005"],  # Employee IDs
    "Project": ["Project 1","Project 2","Project 3","Project 4","Project 5","Project 6","Project 7","Project 8","Project 9","Project 10","Project 11","Project 12","Project 13","Project 14"],  # Project names
    "Cost": [1002000,2000000,4500000,5500000,np.nan,680000,400000,350000,np.nan,300000,2000000,1000000,3000000,200000],  # Project costs with missing values
    "Status": ["Finished","Ongoing","Finished","Ongoing","Finished","Failed","Finished","Failed","Ongoing","Finished","Failed","Ongoing","Finished","Finished"]  # Project status
}

# Convert dictionary into Project DataFrame
Project = pd.DataFrame(project_data)  # Create Project DataFrame

# Create Employee DataFrame data dictionary
employee_data = {
    "ID": ["A001","A002","A003","A004","A005"],  # Employee IDs
    "Name": ["John Alter","Alice Luxemburg","Tom Sabestine","Nina Adgra","Amy Johny"],  # Employee names
    "Gender": ["M","F","M","F","F"],  # Employee gender
    "City": ["Paris","London","Berlin","Newyork","Madrid"],  # Employee city
    "Age": [25,27,29,31,30]  # Employee age
}

# Convert dictionary into Employee DataFrame
Employee = pd.DataFrame(employee_data)  # Create Employee DataFrame

# Create Seniority DataFrame data dictionary
seniority_data = {
    "ID": ["A001","A002","A003","A004","A005"],  # Employee IDs
    "Designation Level": [2,2,3,2,3]  # Designation levels
}

# Convert dictionary into Seniority DataFrame
Seniority = pd.DataFrame(seniority_data)  # Create Seniority DataFrame

# Save all DataFrames into CSV files
Project.to_csv("Project.csv", index=False)  # Save Project CSV
Employee.to_csv("Employee.csv", index=False)  # Save Employee CSV
Seniority.to_csv("Seniority.csv", index=False)  # Save Seniority CSV


# ==========================================
# Task 2 — Replace missing Cost using average with FOR loop
# ==========================================

# Reload Project CSV file
Project = pd.read_csv("Project.csv")  # Read Project CSV

# Calculate average cost ignoring missing values
avg_cost = Project["Cost"].mean()  # Compute mean cost

# Iterate through rows to replace missing values
Project.fillna(avg_cost)  # Replace missing values with average cost

# ==========================================
# Task 3 — Split Name into First Name and Last Name
# ==========================================

# Reload Employee CSV file
Employee = pd.read_csv("Employee.csv")  # Read Employee CSV

# Split Name column into First Name and Last Name
Employee[["First Name", "Last Name"]] = Employee["Name"].str.split(" ", expand=True)  # Split names

# Remove original Name column
Employee.drop("Name", axis=1, inplace=True)  # Drop Name column

# Save the modified Employee DataFrame to CSV
Employee.to_csv("Employee.csv", index=False) # Save the updated Employee CSV


# ==========================================
# Task 4 — Join all DataFrames into Final DataFrame
# ==========================================

# Reload Seniority CSV file
Seniority = pd.read_csv("Seniority.csv")  # Read Seniority CSV

# Merge Project and Employee DataFrames
Final = pd.merge(Project, Employee, on="ID")  # Merge first two tables

# Merge the result with Seniority DataFrame
Final = pd.merge(Final, Seniority, on="ID")  # Merge third table


# ==========================================
# Task 5 — Add Bonus column (5% for Finished projects)
# ==========================================

# Create Bonus column initialized to zero
Final["Bonus"] = 0.0  # Initialize Bonus column as float

# Assign bonus only for finished projects
Final.loc[Final["Status"] == "Finished", "Bonus"] = Final["Cost"] * 0.05  # Calculate 5% bonus


# ==========================================
# Task 6 — Demote designation for failed projects and remove level > 4
# ==========================================

# Reduce designation level by 1 for failed projects
Final.loc[Final["Status"] == "Failed", "Designation Level"] -= 1  # Demote level

# Remove records where designation level exceeds 4
Final = Final[Final["Designation Level"] <= 4]  # Filter rows


# ==========================================
# Task 7 — Add Mr./Mrs. prefix and drop Gender column
# ==========================================

# Add Mr. prefix for male employees
Final.loc[Final["Gender"] == "M", "First Name"] = "Mr. " + Final["First Name"]  # Prefix Mr.

# Add Mrs. prefix for female employees
Final.loc[Final["Gender"] == "F", "First Name"] = "Mrs. " + Final["First Name"]  # Prefix Mrs.

# Drop Gender column
Final.drop("Gender", axis=1, inplace=True)  # Remove Gender column


# ==========================================
# Task 8 — Promote designation level if Age > 29
# ==========================================

# Increase designation level by 1 where age greater than 29
Final.loc[Final["Age"] > 29, "Designation Level"] += 1  # Promote level


# ==========================================
# Task 9 — Calculate total project cost per employee
# ==========================================

# Group data by ID and First Name and sum cost
TotalProjCost = Final.groupby(["ID", "First Name"])["Cost"].sum().reset_index()  # Aggregate cost

# Rename Cost column to Total cost
TotalProjCost.rename(columns={"Cost": "Total cost"}, inplace=True)  # Rename column


# ==========================================
# Task 10 — Print employees whose city contains letter 'o'
# ==========================================

# Filter records where city contains letter 'o'
Employees_with_o = Final[Final["City"].str.contains("o", case=False)]  # Apply condition

# Display final outputs
print("\nFinal DataFrame:\n")  # Print heading
print(Final)  # Show Final DataFrame

print("\nTotal Project Cost DataFrame:\n")  # Print heading
print(TotalProjCost)  # Show total cost table

print("\nEmployees whose city contains letter 'o':\n")  # Print heading
print(Employees_with_o)  # Show filtered employees
```

```python
Final.to_csv("Final.csv", index=False)  # Save Final DataFrame to CSV
Employees_with_o.to_csv("Employees_with_o.csv", index=False)  # Save Employees_with_o DataFrame to CSV
```
